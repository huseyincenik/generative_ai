# Kimi-K2-Thinking

![Hugging Face](https://huggingface.co/front/assets/huggingface_logo.svg)

**Model by:** Moonshot AI  
**Likes:** 1.63k  
**License:** Modified MIT  

---

## 1. Model Overview

**Kimi K2 Thinking** is an advanced open-source thinking model built as an agent capable of multi-step reasoning and dynamic tool invocation. It excels at reasoning, coding, and research tasks by maintaining stable performance over long sequences of tool calls (200–300 steps).  

- **Deep Thinking & Tool Orchestration:** Integrates chain-of-thought reasoning with function calls.  
- **Native INT4 Quantization:** Achieves low-latency inference with 2x speed-up.  
- **Stable Long-Horizon Agency:** Maintains coherent behavior over hundreds of sequential tool invocations.  

---

## 2. Model Architecture

| Attribute | Value |
|-----------|-------|
| Architecture | Mixture-of-Experts (MoE) |
| Total Parameters | 1T |
| Activated Parameters | 32B |
| Layers | 61 |
| Dense Layers | 1 |
| Attention Hidden Dimension | 7168 |
| MoE Hidden Dimension | 2048 per expert |
| Attention Heads | 64 |
| Experts | 384 (8 selected per token) |
| Vocabulary Size | 160K |
| Context Length | 256K |
| Attention Mechanism | MLA |
| Activation Function | SwiGLU |

---

## 3. Evaluation Results

**Reasoning Tasks:**  

| Benchmark | Setting | K2 Thinking | GPT-5 |
|-----------|--------|-------------|-------|
| HLE (Text-only) | no tools | 23.9 | 26.3 |
| HLE | w/ tools | 44.9 | 41.7 |
| AIME25 | w/ python | 99.1 | 99.6 |
| HMMT25 | w/ python | 95.1 | 96.7 |
| IMO-AnswerBench | no tools | 78.6 | 76.0 |

**General Tasks:**  

| Benchmark | Setting | K2 Thinking | GPT-5 |
|-----------|--------|-------------|-------|
| MMLU-Pro | no tools | 84.6 | 87.1 |
| MMLU-Redux | no tools | 94.4 | 95.3 |
| Longform Writing | no tools | 73.8 | 71.4 |

**Agentic Search & Coding Tasks:**  
K2 Thinking consistently outperforms prior models on BrowseComp, Seal-0, and coding benchmarks when tools are enabled.

---

## 4. Features

- **Tool Calling:** Supports dynamic function calls in response to user queries.  
- **Large Context Window:** 256k tokens for long-form reasoning.  
- **Quantization-Aware Training:** Native INT4 for efficient inference.  
- **Flexible Deployment:** Compatible with vLLM, SGLang, and KTransformers.

# <p style="background-color:#FF6961; font-family:newtimeroman;color:#FFF9ED; font-size:150%; text-align:center; border-radius: 15px 50px;"> ✨ Create Tools ✨</p>

<a href="#toc" class="btn btn-warning btn-lg active" role="button" aria-pressed="true" style="color:black;">Content</a>

````python
def get_exchange_rate(currency_pair: str):
    """
    Get the current exchange rate for a currency pair.
    Example: "USD/EUR"
    """
    rates = {
        "USD/EUR": "0.93",
        "EUR/USD": "1.08",
        "USD/TRY": "27.50",
        "EUR/TRY": "28.50"
    }
    return rates.get(currency_pair.upper(), "Rate not available")
````

````python
get_exchange_rate_tool_schema = {
    "type": "function",
    "function": {
        "name": "get_exchange_rate",
        "description": "Get the current exchange rate for a currency pair.",
        "parameters": {
            "type": "object",
            "properties": {
                "currency_pair": {
                    "type": "string",
                    "description": "The currency pair to get the exchange rate for, e.g., USD/EUR."
                }
            },
            "required": ["currency_pair"]
        }
    }
}
````

````python
# Easier way to create the schema

from pydantic import BaseModel, Field

class GetExchangeRateArgs(BaseModel):
    currency_pair: str = Field(..., description="The currency pair to get the exchange rate for. Example: USD/EUR")

exchange_rate_tool_schema = {
    "type": "function",
    "function": {
        "name": "get_exchange_rate",
        "description": "Get the current exchange rate for a currency pair.",
        "parameters": GetExchangeRateArgs.model_json_schema()
    }
}


exchange_rate_tool_schema
````

````python
response = client.chat.completions.create(
    messages=[
        {"role": "user", "content": "what is the exchange rate for USD/EUR?"}
    ],
    tools=[get_exchange_rate_tool_schema],
    tool_choice="auto"
)
````

<a id="create_agent"></a>
# <p style="background-color:#FF6961; font-family:newtimeroman;color:#FFF9ED; font-size:150%; text-align:center; border-radius: 15px 50px;"> ✨ Create Agent ✨</p>

<a href="#toc" class="btn btn-warning btn-lg active" role="button" aria-pressed="true" style="color:black;">Content</a>

````python
import json

class Agent:
    def __init__(self, client: InferenceClient, system: str = "", tools: list = None) -> None:
        self.client = client
        self.system = system
        self.messages: list = []
        self.tools = tools if tools is not None else []
        if self.system:
            self.messages.append({"role": "system", "content": system})

    def __call__(self, message=""):
        if message:
            self.messages.append({"role": "user", "content": message})

        final_assistant_content = self.execute()

        if final_assistant_content:
            self.messages.append({"role": "assistant", "content": final_assistant_content})

        return final_assistant_content

    def execute(self):
        # Keep looping until the model provides a final text response (not tool calls)
        while True:
            completion = self.client.chat.completions.create(
                messages=self.messages,
                tools=self.tools,
                tool_choice="auto" # Allow the model to decide whether to call tools
            )

            response_message = completion.choices[0].message

            if response_message.tool_calls:
                self.messages.append(response_message)

                tool_outputs = []
                for tool_call in response_message.tool_calls:
                    function_name = tool_call.function.name
                    function_args = json.loads(tool_call.function.arguments)

                    # Execute the tool
                    if function_name in globals() and callable(globals()[function_name]):
                        function_to_call = globals()[function_name]
                        executed_output = function_to_call(**function_args)
                        tool_output_content = str(executed_output) # Ensure output is string
                        print(f"Executing tool: {function_name} with args {function_args}, Output: {tool_output_content[:500]}...") # Debug print

                    tool_outputs.append(
                        {
                            "tool_call_id": tool_call.id,
                            "role": "tool",
                            "name": function_name,
                            "content": tool_output_content,
                        }
                    )

                self.messages.extend(tool_outputs)

            else:
                return response_message.content
````

````python
agent = Agent(
    client=client,
    system="You are a helpful assistant that can answer questions using the provided tools.",
    tools=[get_exchange_rate_tool_schema]
)


response = agent("what is the exchange rate for USD/EUR?")
print(response)
````
