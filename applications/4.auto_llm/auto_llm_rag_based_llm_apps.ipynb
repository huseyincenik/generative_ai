<div style="background-color:#3498DB; padding:10px; border-radius: 10px; box-shadow: 2px 2px 4px rgba(0, 0, 0, 0.5);">
    <p style="font-weight: bold; color: white;">AutoQueryEngine: A Context-Based Query Answering System

`AutoQueryEngine` is a tool designed to enhance query answering by leveraging document context. It aims to generate `relevant and accurate responses` based on a given set of information. Here's how it works:

- `system prompt`: Guides the AI assistant to focus on providing answers that are `relevant` and `accurate`.
- `query_wrapper_prompt`: A `template` that structures the query and incorporates the provided document context for generating responses.
- `context-aware answers`: By grounding the response in the specific context, it ensures that the answer is tailored to the user's question, offering better `precision`.

This approach ensures that queries are answered with a high degree of `relevance` and `accuracy` by relying heavily on the provided context.</p>
</div>


````python
query_engine = AutoQueryEngine.from_defaults(documents)
````

````python
system_prompt = "You are an AI assistant that helps users find the most relevant accurate answers to their questions."

query_wrapper_prompt = """

The document information is below.
----------------------------------
{context_str}
----------------------------------
Using the document information and mostly relying on it,
answer the query.
Query: {query_str}
Answer:
"""
````

<a href="#part_1" class="btn btn-danger btn-lg active" role="button" aria-pressed="true" style="color:white;">Part - 1</a>

<div style="background-color:#3498DB; padding:10px; border-radius: 10px; box-shadow: 2px 2px 4px rgba(0, 0, 0, 0.5);">
    <p style="font-weight: bold; color: white;">This code snippet defines a framework for generating answers to user queries based on provided document context. 
        
-The `system prompt` helps the AI assistant focus on finding `relevant and accurate information`, while the `query_wrapper_prompt` serves as a `template` for incorporating context into the query-answering process. 
- Essentially, it helps ensure that `responses` are grounded in the specific context, leading to more `accurate` and `relevant answers`.</p>
</div>


````python
enable_cost_calculator = True
````

````python
model = "gpt-4o-mini"
````

````python
vector_store_type = "LanceDBVectorStore"
````

````python
uri = "tmp/lancedb"
table_name = "vectors"
````

````python
chunk_size = 1024
````

````python
similarity_top_k = 5
````

````python
llm_params = {"model" : model}
````

````python
vector_store_params = {"vector_store_type": vector_store_type, "uri": uri, "table_name": table_name}
````

````python
service_context_params = {"chunk_size": chunk_size}
````

````python
query_engine_params = {"similarity_top_k": similarity_top_k}
````

````python
query_engine_new = AutoQueryEngine.from_defaults(documents = documents,
                                                 system_prompt = system_prompt,
                                                 query_wrapper_prompt = query_wrapper_prompt,
                                                 enable_cost_calculator = enable_cost_calculator,
                                                 llm_params = llm_params,
                                                 vector_store_params = vector_store_params,
                                                 service_context_params = service_context_params,
                                                 query_engine_params = query_engine_params)
````

<div style="background-color:#3498DB; padding:10px; border-radius: 10px; box-shadow: 2px 2px 4px rgba(0, 0, 0, 0.5);">
    <p style="font-weight: bold; color: white;">The code snippet defines an instantiation of a **query engine** using `AutoQueryEngine`. Here's a breakdown of the key components:

- **`documents`**: The collection of documents that will be used for querying.
- **`system_prompt`**: The pre-configured prompt guiding the AI to find **relevant and accurate** information.
- **`query_wrapper_prompt`**: A template that incorporates the document context into the query-answering process.
- **`enable_cost_calculator`**: Optionally enables a feature to track the cost associated with queries.
- **`llm_params`**: Parameters for the **Language Model** (LLM), configuring its behavior during queries.
- **`vector_store_params`**: Parameters for the **vector store**, optimizing document retrieval.
- **`service_context_params`**: Context parameters that help integrate external services or configurations.
- **`query_engine_params`**: Additional parameters controlling the query engine's functionality.

This setup enables an efficient and **context-aware querying process, ensuring responses are tailored to the **provided documents**.
</p>
</div>

This guide walks you through setting up AutoLLM, importing libraries, reading documents, and querying using AutoQueryEngine—all wrapped up in a user-friendly workflow.

AutoLLM allows you to integrate LLMs with RAG capabilities effortlessly. By leveraging libraries like `gradio` and `AutoQueryEngine`, you can create AI-powered applications in just a few lines of code. Key steps include:
- Importing essential libraries.
- Reading files or GitHub repositories as documents.
- Configuring and initializing the AutoQueryEngine.
